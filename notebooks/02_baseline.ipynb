# 02 — Baselines (Random Forest / SVM)

Strong classical baselines on the same HVG features, evaluated leave-one-dataset-out.
Primary metrics are **PR-AUC / ROC-AUC** (threshold-free), because the ≈18:1 imbalance
plus cross-platform score shift makes accuracy / F1 misleading.


```python
from pathlib import Path
import yaml

for f in ["results/lodo_results.yaml", "results/tumor_normal_results.yaml",
          "results/tn_transformer_results.yaml"]:
    p = Path(f)
    if p.exists():
        print("=" * 70)
        print(f)
        print("=" * 70)
        print(yaml.safe_load(p.read_text(encoding="utf-8")))
        print()
```

## LODO results (threshold-free)

| Model | PR-AUC (mean ± std) | ROC-AUC (mean ± std) |
|---|---|---|
| Random Forest | 0.9961 ± 0.0025 | 0.9486 ± 0.0250 |
| SVM (RBF) | 0.9986 ± 0.0012 | 0.9776 ± 0.0154 |
| Gene Transformer | 0.9963 ± 0.0029 | 0.9637 ± 0.0192 |

The Transformer reaches PR-AUC parity with the SVM and beats RF on ROC-AUC, while
remaining end-to-end differentiable (the property the SHAP/attention analysis needs).
