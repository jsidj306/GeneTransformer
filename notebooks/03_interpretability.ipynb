# 03 — Interpretability (SHAP + Attention + Enrichment)

Two inference-only views of *why* the Transformer says tumor, plus GO/KEGG/Reactome
enrichment of the cross-fold consensus genes. No retraining — run on CPU from the
local checkpoints.


```python
from pathlib import Path
import pandas as pd

top = pd.read_csv("results/tn_interpretability_top_genes.csv")
top = top.sort_values("combined_rank_score")
print("consensus genes (top 20, rank-based, n_folds=3):")
print(", ".join(top["gene"].head(20).astype(str)))
```

```python
enr = pd.read_csv("results/enrichment_results.csv")
sig = enr[enr["significant"]]
print("FDR-significant terms (adj p < 0.05):")
if sig.empty:
    print("  (none)")
else:
    for _, r in sig.iterrows():
        print(f"  {r['library']:12s} {r['term']}  (adj p={r['adj_p_value']:.2g}, overlap={r['overlap']})")
```
